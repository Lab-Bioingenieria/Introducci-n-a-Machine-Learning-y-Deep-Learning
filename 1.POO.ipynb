# Introducción a la Programación Orientada a Objetos (POO) en Biomedicina

¡Hola! Bienvenido a este blog interactivo enfocado al sector médico y biológico. Aquí aprenderemos los conceptos básicos de la Programación Orientada a Objetos (POO) utilizando Python, modelando la vida real clínica.

La POO es un paradigma que se basa en el concepto de **"objetos"**. Así como en un hospital interactúan pacientes, doctores y máquinas de diagnóstico (cada uno con su estado o atributos funcionales), en POO modelamos exactamente de esa manera.

## 1. Clases y Objetos (El Paciente)

Imagina que una **Clase** es la ficha médica en blanco, el molde universal de "Cómo debe ser registrado un enfermo", y un **Objeto** es el paciente físico ya ingresado, por ejemplo, "María, 45 años".

```python
class Paciente:
    # Atributo de clase (compartido por todas las personas del hospital)
    especie = "Homo sapiens"

    # Constructor (__init__), que se llama al registrar a un paciente nuevo en el sistema
    def __init__(self, nombre, edad, grupo_sanguineo):
        self.nombre = nombre        # Atributo: nombre del paciente
        self.edad = edad            # Atributo: su edad biológica
        self.sangre = grupo_sanguineo # Atributo: su tipo de sangre

    # Método (Una acción que el paciente puede emitir asígnado a su propia ficha)
    def reportar_sintomas(self, sintoma):
        return f"El paciente {self.nombre} (Sangre {self.sangre}) reporta: {sintoma}."

# Ingresando pacientes de urgencias (Instanciando objetos)
paciente_1 = Paciente("Carlos Gómez", 32, "O+")
paciente_2 = Paciente("Elena Torres", 58, "A-")

print(f"Ingreso: {paciente_1.nombre}, Edad: {paciente_1.edad}")
print(paciente_2.reportar_sintomas("Dificultad severa para respirar"))
```

## 2. Los 4 Pilares de la POO (Aplicados a Salud)

Ahora veremos qué hace tan poderoso a este modelo en proyectos de Bioinformática y sistemas hospitalarios aplicando: **Encapsulamiento, Herencia, Polimorfismo y Abstracción**.

### 2.1 Encapsulamiento (Confidencialidad Médica)

Así como el diagnóstico grave de un paciente o su ADN es confidencial (HIPAA), en el código evitamos que cualquiera modifique directamente variables críticas usando `__` (atributos privados). Si un doctor quiere modificar un diagnóstico, debe usar una función de seguridad.

```python
class ExpedienteClinico:
    def __init__(self, id_paciente, diagnostico_inicial):
        self.id = id_paciente
        self.__diagnostico = diagnostico_inicial # ¡Dato privado y protegido!
        self.__medicamento = "Ninguno"

    # Método seguro para cambiar el diagnóstico sólo si hay autorización
    def actualizar_diagnostico(self, clave_doctor, nuevo_diagnostico):
        if clave_doctor == "DOC-123": # Validación de acceso
            self.__diagnostico = nuevo_diagnostico
            print("Diagnóstico médico actualizado con éxito.")
        else:
            print("ALERTA: Acceso denegado. No tiene permisos para editar la enfermedad de este paciente.")

    def leer_expediente(self):
        return f"Paciente {self.id} -> Estado actual: {self.__diagnostico}"

mi_expediente = ExpedienteClinico("P-9981", "Tos leve")

mi_expediente.actualizar_diagnostico("DOC-000", "Neumonía Bácteriana") # ¡Falla!
mi_expediente.actualizar_diagnostico("DOC-123", "Neumonía Bácteriana") # ¡Aprobado!
print(mi_expediente.leer_expediente())

# Si intentas hackear accediendo directo a '__diagnostico', dará error Python:
# print(mi_expediente.__diagnostico)
```

### 2.2 Herencia (Biología Celular)

Toda célula parte de un diseño maestro (Célula Madre), pero las células "Hijas" (Eritrocitos, Macrófagos) heredan núcleos y la capacidad de morir y nacer, añadiendo sus propias características sin repetir el código biológico desde cero.

```python
class Celula: # Clase Padre (Metabolismo base de cualquier célula)
    def __init__(self, tipo_tejido):
        self.tipo_tejido = tipo_tejido
        self.viva = True

    def consumir_glucosa(self):
        print("Consumiendo glucosa para generar ATP...")

# La célula inmunológica hereda la capacidad de comer glucosa del padre
class Macrofago(Celula):
    def __init__(self):
        super().__init__("Tejido Inmunitario") # Llamamos al constructor del padre
        self.patogenos_fagocitados = 0

    def fagocitar(self, bacteria):
        self.patogenos_fagocitados += 1
        return f"¡El macrófago ha devorado a la {bacteria}! Cuenta: {self.patogenos_fagocitados}"

linfocito_soldado = Macrofago()

linfocito_soldado.consumir_glucosa() # Método heredado
print(linfocito_soldado.fagocitar("E. Coli")) # Método propio de su especialización
```

### 2.3 Polimorfismo (Respuesta Farmacológica)

Polimorfismo es tener fármacos distintos (Paracetamol, Quimioterapia, Antibiótico). A todos les puedes llamar `.administrar()`, pero la acción biológica (*la máquina farmacológica interna*) actuará de maneras completamente diferentes.

```python
class Antibiotico:
    def administrar(self):
        return "Rompiendo la pared celular de las bacterias."

class Antiviral:
    def administrar(self):
        return "Bloqueando la replicación del ARN viral dentro de la célula hésped."

class Inmunosupresor:
    def administrar(self):
        return "Disminuyendo la acción inflamatoria del sistema inmune."

# Tratamiento múlti-droga para un paciente grave
cocktail_medico = [Antibiotico(), Antiviral(), Inmunosupresor()]

for medicina in cocktail_medico:
    # POLIMORFISMO: Misma orden, resultados moleculares dispares en cadena
    print(f"-> Acción: {medicina.administrar()}")
```

### 2.4 Abstracción (Equipos de Diagnóstico - 'Caja Negra')

Para operar la máquina de Rayos X o Resonancia Magnética, sólo necesitas meter al paciente y presionar "Escanear" (`.escanear()`). No necesitas conocer la ecuación abstracta de física radiológica que está calculando internamente. Oculta la complejidad.

```python
from abc import ABC, abstractmethod

# Regulación obligatoria bio-ética. Toda máquina DEBE tener la interfaz Escanear y Detenerse por seguridad.
class DispositivoMedico(ABC):
    @abstractmethod
    def escanear(self):
        pass

class Tomografia(DispositivoMedico):
    def escanear(self):
        return "Generando 500 cortes transversales por rayos X del interior del cráneo... Result: Tumor Detectado"

class Electrocardiograma(DispositivoMedico):
    def escanear(self):
        return "Midiendo impulsos eléctricos del ventrículo cada milisegundo... Result: Arritmia leve"

mi_tomografo = Tomografia()
mi_electro = Electrocardiograma()

print("Abstracción pura: El doctor sólo aprieta un botón de 'escanear'.")
print(f"TOMÓGRAFO: {mi_tomografo.escanear()}")
print(f"ELECTROCARDIOGRAMA: {mi_electro.escanear()}")
```

- La **Herencia** modela biológicamente cuándo una célula hija adquiere habilidades innatas de su célula madre.
